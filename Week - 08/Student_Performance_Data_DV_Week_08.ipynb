Grouped Bar Chart - Parental Education

```python
parental_scores = df.groupby('parental level of education')[score_columns].mean()
parental_scores.plot(kind = 'bar',figsize = (12,6))
plt.title('Average Exam Scores by Parental Level of Education')
plt.xlabel('Parental Level of Education')
plt.ylabel('Average Score')
plt.xticks(rotation = 45)
plt.legend(title = 'Subject')
plt.tight_layout()
plt.show()
```

```python
plt.title("Score Variance Based on Test Preparation")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["math score", "reading score", "writing score"]
)
plt.show()
```

```python
scores = df[
    ["math score", "reading score", "writing score"]
]
corr = scores.corr()
```

```python
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
```

```python
plt.title("Correlation Between Math, Reading and Writing Scores")
plt.show()
```
